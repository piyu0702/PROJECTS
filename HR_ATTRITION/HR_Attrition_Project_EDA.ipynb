# HR Employee Attrition Analysis using Python

## Introduction



- Employee attrition is one of the major challenges faced by organizations.
- High employee turnover can increase recruitment costs, reduce productivity, and affect workforce stability.
- This project performs Exploratory Data Analysis (EDA) on the IBM HR Employee Attrition dataset to identify the key factors that influence employee turnover. - The dataset contains 1,470 employee records and 35 features including age, salary, department, job role, overtime, and satisfaction levels.

Dataset Source: IBM HR Analytics Employee Attrition Dataset (available on Kaggle)

## Project Objective

#### The objective of this project is to analyze employee attrition data and identify the major factors responsible for employee turnover using Python-based exploratory data analysis techniques.


## Problem Statement

- The overall attrition rate in this dataset is 16.1%, meaning approximately 1 in 6 employees left the organization.
- This project aims to identify which specific factors — such as overtime, salary, age, job role, and satisfaction — are most strongly associated with employee attrition, so that HR teams can take targeted action to improve retention.

## Attrition Distribution Analysis

The `countplot()` function is used to visualize employee attrition frequency.

```python
plt.figure(figsize=(6,4))
num = sns.countplot(x='Attrition', data=df)

for container in num.containers:
    num.bar_label(container)

plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition")
plt.ylabel("Employee Count")
plt.show()
```

#### Information Derived
- The chart shows the number of employees who stayed and left the organization.
- Most employees remained in the company, while a smaller portion left the organization.
- This indicates employee attrition exists but is comparatively lower than employee retention.
- Understanding attrition distribution helps organizations evaluate workforce stability.
Out of 1,470 employees:
- 1,233 employees (83.9%) stayed in the organization.
- 237 employees (16.1%) left the organization.

## Monthly Income vs Attrition

Monthly Income vs Attrition Analysis

```python
plt.figure(figsize=(8,5))

avg_income = df.groupby('Attrition')['MonthlyIncome'].mean()

num = avg_income.plot(kind='bar')

for container in num.containers:
    num.bar_label(container)

plt.title("Average Monthly Income by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Average Monthly Income")
plt.xticks(rotation=0)
plt.show()
```

#### Information Derived
- The chart compares the average monthly income of employees who stayed and employees who left the organization.
- Employees who stayed in the company have an average monthly income of approximately **6,832**, while employees who left have an average monthly income of approximately **4,787**.
- Employees who left earned about 30% less than employees who stayed.
- This confirms that lower salary is strongly linked to higher attrition.
